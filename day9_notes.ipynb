# Day 3 数学基础：方差、标准差与运算性质

## 1. 方差 (Variance)
**物理意义**：衡量随机变量 $X$ 取值的分散程度（距离期望值的平均偏离平方）。
*   **定义公式**：$$Var(X) = E[(X - \mu)^2]$$
    *   其中 $\mu = E[X]$
*   **计算公式（常用推导）**：$$Var(X) = E[X^2] - (E[X])^2$$

## 2. 标准差 (Standard Deviation)
**物理意义**：方差的正平方根。与随机变量 $X$ 具有**相同的量纲（单位）**，在工程和数据描述中更直观。
*   **公式**：$$\sigma = \sqrt{Var(X)}$$

## 3. 核心运算性质 (极其重要)
*假设 $X, Y$ 为随机变量，$c$ 为常数。*

### 期望的性质 (线性性质)
*   **常数**：$E[c] = c$
*   **乘常数**：$E[cX] = cE[X]$
*   **加减法**：$E[X \pm Y] = E[X] \pm E[Y]$ (无论 $X, Y$ 是否独立均成立)

### 方差的性质
*   **常数**：$Var(c) = 0$ (常数没有波动)
*   **乘常数**：$Var(cX) = c^2 Var(X)$ (注意：**方差会被放大 $c^2$ 倍**)
*   **平移**：$Var(X + c) = Var(X)$ (整体平移不改变分散程度)
*   **加法（独立前提）**：**如果 $X$ 和 $Y$ 相互独立**，则 $Var(X \pm Y) = Var(X) + Var(Y)$
    *   *注意：如果是相减 $Var(X-Y)$，方差依然是相加，因为不确定性是叠加的。*

## 4. 深度学习关联：标准化 (Standardization / Z-score)
**目的**：将任意分布转化为均值为 $0$，标准差为 $1$ 的标准分布，是 **Batch Normalization** 的核心基础。
*   **公式**：$$Z = \frac{X - \mu}{\sigma}$$
*   **性质验证**：
    *   $E[Z] = E[\frac{X - \mu}{\sigma}] = \frac{1}{\sigma}(E[X] - \mu) = \frac{1}{\sigma}(\mu - \mu) = 0$
    *   $Var(Z) = Var(\frac{X - \mu}{\sigma}) = \frac{1}{\sigma^2} Var(X) = \frac{1}{\sigma^2} \sigma^2 = 1$

# Day 3 理论通关速查表：反向传播四大基本公式

在推导反向传播时，我们统一引入一个核心变量：**误差项 $\delta^{(l)}$**。
它的定义是损失 $L$ 对第 $l$ 层**激活前状态 $z^{(l)}$** 的偏导数：
$$\delta^{(l)} = \frac{\partial L}{\partial z^{(l)}}$$

---

## 公式 1：输出层的误差 (BP1)
描述最后一层（第 $L$ 层）的误差是如何产生的。
$$\delta^{(L)} = \frac{\partial L}{\partial a^{(L)}} \odot \sigma'(z^{(L)})$$
* **物理直觉**：Loss 对最终输出的偏导，乘上最后一层激活函数的导数（$\odot$ 表示逐元素相乘）。

## 公式 2：误差的反向传递 (BP2 - 链式核心)
描述第 $l$ 层的误差，如何由下一层（第 $l+1$ 层）的误差推导过来。
$$\delta^{(l)} = ((W^{(l+1)})^T \delta^{(l+1)}) \odot \sigma'(z^{(l)})$$
* **物理直觉**：上游的误差 $\delta^{(l+1)}$ 乘以上游权重矩阵的转置（按权重分配责任），再经过当前层的激活函数门控开关。

## 公式 3：偏置的梯度 (BP3)
描述损失 $L$ 对当前层（第 $l$ 层）偏置 $b^{(l)}$ 的偏导。
$$\frac{\partial L}{\partial b^{(l)}} = \delta^{(l)}$$
* **物理直觉**：偏置的梯度，就直接等于当前层的误差项。

## 公式 4：权重的梯度 (BP4)
描述损失 $L$ 对当前层（第 $l$ 层）权重 $W^{(l)}$ 的偏导。
$$\frac{\partial L}{\partial W^{(l)}} = \delta^{(l)} (a^{(l-1)})^T$$
* **物理直觉**：当前层的误差项，乘上来自上一层的输入 $a^{(l-1)}$ 的转置（外积组合），维度完美对齐。
