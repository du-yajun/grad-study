##  一、 理论核心：反向传播 (BP) 到底在算什么？

反向传播本质上是在利用微积分的**链式法则 (Chain Rule)**，高效地计算损失函数 (Loss) 对网络中每一个参数 (Weight & Bias) 的梯度。

整个过程分为两个阶段：

### 1. Forward Pass (前向传播)
* **目的**：计算并保存所有的**激活值 (Activation)**。
* **过程**：数据 $x$ 从输入层出发，顺着网络一层层计算，直到得出最终的预测结果 $y$ 和 Loss。
* **注意**：这个时候先不管梯度，只管沿着网络往前算，并把中间结果存下来，后面要用。

### 2. Backward Pass (反向传播)
* **目的**：计算所有参数的梯度 $\frac{\partial L}{\partial w}$。
* **过程**：假设网络末端引入一个微小的扰动，顺着计算图**反向**退回去。
* **核心精髓 (为什么高效)**：
  反向传播极其聪明的地方在于 **“复用”**。计算某一层梯度时，直接乘上上一层已经算好的梯度，以及 Forward Pass 中保存的激活值。避免了从头到尾重新展开多项式的海量重复计算。

## 1. 期望 (Expected Value) 的核心定义
期望本质上是随机变量取值的**加权平均值**（以概率作为权重）。在机器学习中，它常用来衡量模型预测的平均水平。

**数学公式：**
设 $X$ 是一个随机变量，$p(x)$ 或 $f(x)$ 是对应的概率/概率密度。
*   **离散型：**
    $$E[X] = \sum x_i \cdot P(X = x_i)$$
    *(所有可能发生的结果 $x_i$，乘以它们发生的概率，然后全部加起来)*
*   **连续型：**
    $$E[X] = \int_{-\infty}^{\infty} x \cdot f(x) dx$$

---

## 2. 函数的期望 (Expected Value of a Function)
如果我们关心的不是 $X$ 本身，而是 $X$ 的某个函数 $g(X)$（比如 $g(X) = X^2$），期望的求法依然是用概率做加权：
*   **离散型：**
    $$E[g(X)] = \sum g(x_i) \cdot P(X = x_i)$$
*   **连续型：**
    $$E[g(X)] = \int_{-\infty}^{\infty} g(x) \cdot f(x) dx$$

> 💡 **AI 实战意义：** 在训练模型时，我们算 Loss 的过程，其实就是在算整个数据集上损失函数值的期望！

---

## 3. 期望的线性性质 (Linearity of Expectation) ⭐⭐⭐
这是期望**最重要**的性质！它说明无论随机变量之间是否相互独立，期望都可以进行线性拆分。

**核心公式：**
1.  **常数可提取：** 
    $E[aX + b] = aE[X] + b$  *(其中 $a, b$ 为常数)*
2.  **期望求和可拆分（极其重要）：**
    $E[X + Y] = E[X] + E[Y]$

> 📌 **大白话翻译：** 整体的期望，等于各个部分期望的和。这在推导复杂模型的总体误差时，能帮我们把极其复杂的公式拆解成好算的小块。
